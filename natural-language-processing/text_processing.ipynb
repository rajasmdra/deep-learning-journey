# CASE FOLDING
Change text from uppercase to lowercase

```python
real_text = "Ini Adalah Contoh Teks yang Akan Dikonversi Menjadi Lowercase."

lowered_text = real_text.lower()

print("Real text:", real_text)
print("Lowered text:", lowered_text)
```

# REMOVE SPECIAL CHARACTER

```python
def remove_char(text):
    text_without_char = ''.join([char for char in text if char.isalpha() or char.isdigit() or char.isspace()])
    return text_without_char

text_with_char = "Hari ini cuaca sangat cerah! Angin bertiup kencang, dan ada beberapa awan di langit. Bagaimana dengan cuaca besok?"

text_without_char = remove_char(text_with_char)

print('Text with char:', text_with_char)
print('Text without char:', text_without_char)
```

# REMOVE DIGIT

```python
def remove_digit(text):
    text_without_digit = ''.join([char for char in text if not char.isdigit()])
    return text_without_digit

text_with_digit = "Ini adalah contoh teks dengan angka 12345 yang akan dihapus."

text_without_digit = remove_digit(text_with_digit)

print("Text with digit:", text_with_digit)
print("Text without digit:", text_without_digit)
```

# REMOVE DIGIT WITH REGEX

```python
import re

def remove_irrelevant_digit(text):
    irrelevant_digit_pattern = r"\b(?:\d{1,3}[-\.\s]?)?(?:\d{3}[-\.\s]?)?\d{4,}\b" # DELETE PHONE NUMBER ONLY
    # irrelevant_digit_pattern = r"\b\d+\b" # DELETE ALL DIGIT
    result = re.sub(irrelevant_digit_pattern, "", text)
    return result.strip()

sentence = "Di sini ada 3 nomor rumah yaitu: 123, 456, dan 789. Silakan hubungi 081234567890 untuk informasi lebih lanjut."

result_sentence = remove_irrelevant_digit(sentence)

print("Sentence with digit:", sentence)
print("Sentence without digit:", result_sentence)
```

# REMOVE WHITESPACE

```python
text_with_whitespace = " Ini adalah contoh kalimat dengan spasi. "
text_after_strip = text_with_whitespace.strip()
text_without_whitespace = text_after_strip.replace(" ", "")

print("Text with whitespace:", text_with_whitespace)
print("Text after strip:", text_after_strip)
print("Text without whitespace:", text_without_whitespace)
```

# STOPWORD REMOVAL

## NLTK

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt_tab')

text = "Perekonomian Indonesia sedang dalam pertumbuhan yang membanggakan."

word_tokens = word_tokenize(text)

stopwords_indonesia = set(stopwords.words('indonesian'))

important_text = [kata for kata in word_tokens if kata.lower() not in stopwords_indonesia]

text_without_stopwords = ' '.join(important_text)

print("Real text:", text)
print("Text without stopwords:", text_without_stopwords)
```

## SASTRAWI

```python
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from nltk.tokenize import word_tokenize

factory = StopWordRemoverFactory()
stopwords_sastrawi = factory.get_stop_words()

text = "Perekonomian Indonesia sedang dalam pertumbuhan yang membanggakan."

word_tokens = word_tokenize(text)

important_text = [kata for kata in word_tokens if kata.lower() not in stopwords_sastrawi]

text_without_stopwords = ' '.join(important_text)

print("Real text:", text)
print("Text without stopwords:", text_without_stopwords)
```

# TOKENIZING

## WORD TOKENIZING

```python
from nltk.tokenize import word_tokenize
text = "Ini adalah contoh tokenisasi kata dalam pemrosesan teks."

tokens = word_tokenize(text)
print(tokens)
```

# SENTENCE TOKENIZING

```python
from nltk.tokenize import sent_tokenize

text = "Ini adalah contoh tokenisasi kalimat. Apakah ini kalimat kedua? Ya, ini kalimat ketiga!"
sentences = sent_tokenize(text)
print(sentences)
```

## PHRASE TOKENIZING

```python
from nltk.tokenize import TreebankWordTokenizer

text = "Pemrosesan teks adalah cabang ilmu komputer yang berfokus pada pengolahan teks dan dokumen."
tokenizer = TreebankWordTokenizer()
phrases = tokenizer.tokenize(text)
word = word_tokenize(text)
print(phrases)
print(word)
```

## RULE BASED TOKENIZING

```python
import re

text = "Pertama, kita perlu menyiapkan bahan-bahan yang diperlukan."
tokens = re.findall(r'\w+|\d+', text)
print(tokens)
```

## MODEL BASED TOKENIZING

```python
text = "Ini adalah contoh tokenisasi berbasis model."
tokens = text.split()
print(tokens)
```

# STEMMING 
Change active verbs into base verbs.

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["running", "runs", "runner", "ran", "easily", "fairness", "better", "best", "cats", "cacti", "geese", "rocks", "oxen"]

for word in words:
    stemmed_word = stemmer.stem(word)
    print(f"{word} - {stemmed_word}")
```

# LEMMATIZATION

```python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

words = ["running", "runs", "runner", "ran", "easily", "fairness", "better", "best", "cats", "cacti", "geese", "rocks", "oxen"]

for word in words:
    lemma_word = lemmatizer.lemmatize(word)
    print(f"{word} - {lemma_word}")
```
